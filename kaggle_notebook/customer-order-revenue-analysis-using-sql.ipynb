# **Business Questions (ASK)**

The objective of this analysis is to understand customer purchasing behavior and overall revenue performance in an e-commerce platform. Specifically, the analysis aims to answer the following questions:

* How many unique customers are active on the platform?
* What proportion of customers are one-time buyers versus repeat buyers?
* How is revenue distributed across customers?
* Is revenue concentrated among a small group of high-value customers?
* How do orders and revenue change over time?
* Are there clear peak periods or seasonal patterns in customer purchases?
* Is revenue growth driven by higher order volume or higher average order value (AOV)?

# **Dataset & Preparation (PREPARE)**

This project uses an e-commerce transactional dataset containing information about customers, orders, and order-level items. The following tables were used for the analysis:

* **customers**: Contains customer identifiers, including customer_id and customer_unique_id, which represents a unique customer across multiple orders.

* **orders**: Stores order-level information such as order_id, customer_id, order status, and the timestamp when an order was placed (order_purchase_timestamp).

* **order_items**: Contains item-level transaction data, including product prices associated with each order.

For this analysis, revenue was defined as the sum of item prices (order_items.price). Shipping and other auxiliary costs were excluded to focus on pure product revenue.
The order purchase timestamp (order_purchase_timestamp) was used as the primary time dimension for all time-based analyses, such as monthly trends and seasonality.

Before starting the analysis, basic data validation was performed to confirm:

* The number of unique customers in the dataset.

* That all customers in the dataset had placed at least one order.

* The relationship between customer_id and customer_unique_id, ensuring customer-level analysis was performed at the correct granularity.

# **Data Processing (PROCESS)**

**SQL** was used as the primary tool for data processing and transformation. To ensure clarity and maintainability, the analysis followed a step-by-step approach, breaking complex logic into smaller, interpretable components using **Common Table Expressions (CTEs)**.

The main processing steps included:

**Customer-level aggregation**:
Orders were grouped at the customer_unique_id level to correctly represent individual customers, avoiding duplication caused by multiple customer_id records for the same person.

**Customer segmentation**:
Customers were classified into one-time and repeat buyers based on the number of distinct orders placed. This segmentation enabled comparison of customer behavior and revenue contribution.

**Revenue calculation**:
Revenue was calculated using the sum of item prices from the order_items table. Aggregations were performed at different levels (customer-level and monthly-level) depending on the analysis objective.

**Time-based transformations**:
The order purchase timestamp was converted into a year-month format (YYYY-MM) to support monthly trend analysis and seasonality detection.

**Use of window functions**:
Ranking functions were applied to identify top customers by revenue and to evaluate revenue concentration, supporting Pareto-style analysis.

This structured processing approach ensured that each metric—customer counts, revenue totals, trends, and averages—was derived consistently and transparently.

**3. Revenue Distribution & Concentration (Pareto Analysis)**

**Purpose**
To understand whether platform revenue is driven by a small group of high-value customers or broadly distributed across the customer base, a revenue concentration analysis was performed.

**SQL Logic**

Customer-level revenue was calculated by aggregating total item prices across all orders placed by each unique customer. Customers were then ranked based on total revenue contribution to evaluate concentration effects.

![sqlimage02](https://lh3.googleusercontent.com/u/0/d/1Arh8BGX2443b93x9dChpBD-4i1jxUsmV)


![furthervalidate](https://lh3.googleusercontent.com/u/0/d/1YwCj954mujOay_HR-VxMxBD5FnNcCM_i)

**Insight**

Revenue contribution is highly fragmented across customers:

Top 10 customers account for only ~0.5% of revenue

Top 100 customers account for just ~2.45% of revenue

This indicates that the platform does not rely on a small group of high-value “whale” customers. Instead, revenue is generated through a large number of low-to-moderate value transactions.

**Business Interpretation**:
While this reduces dependency risk on individual customers, it also suggests limited upside from customer concentration strategies. Growth opportunities may instead lie in improving customer retention or increasing average order value.

**5. Average Order Value (AOV) Trend**

**Purpose**

While revenue and order volume trends explain how much business activity occurs over time, Average Order Value (AOV) helps explain why revenue changes.
AOV measures how much customers spend per order on average and indicates whether growth is driven by higher spending or increased transaction volume.

> AOV = Total Revenue / Total Orders

SQL Logic: Monthly AOV Calculation

Monthly AOV was calculated by first aggregating total revenue and total orders per month and then dividing revenue by order count.
![sqlimage06](https://lh3.googleusercontent.com/u/0/d/1KZSZfjVnS8LCp4k-hwOjmay6d8-9PoPW)

**Result Interpretation: AOV Trend**

The AOV trend remains relatively stable across months, with no strong upward or downward trajectory.

![monthlyaverageordervalue](https://lh3.googleusercontent.com/u/0/d/1rGj3xpVkTSgMYQJmMQdFr3ItWvw22koy)

**Insight**: 
Despite fluctuations in total revenue and order volume, customers do not appear to be spending significantly more per order over time.

**Business Interpretation**

* Revenue growth is not driven by increasing basket size

* Performance improvements are primarily due to changes in order volume

* The platform operates on a high-volume, low-ticket sales model

This reinforces earlier findings from customer and revenue concentration analyses, where revenue was shown to be broadly distributed rather than concentrated among high-spending customers.

**Trend Chapter Summary**

* Monthly revenue and order volume fluctuate over time

* Clear peak periods and seasonality exist

* AOV remains largely stable

* Growth is driven by transaction volume rather than increased spending per order


# **Key Findings (SHARE)**

* The customer base is overwhelmingly composed of one-time buyers, with approximately 97% of customers placing only a single order. Customer retention is low, and repeat purchasing behavior is limited.

* Although repeat customers generate higher average revenue per customer, their small population means they do not significantly influence total platform revenue.

* Revenue is highly fragmented across customers.

    * The top 10 customers contribute only ~0.5% of total revenue.

    * Even the top 100 customers account for just ~2.45% of revenue. This indicates that the platform does not rely on a small group of high-value customers.

* Monthly revenue and order trends show clear fluctuations and peak periods, rather than steady linear growth, highlighting demand-driven variability.

* Revenue trends closely track order volume trends, suggesting that changes in revenue are primarily driven by the number of orders placed, not by customers spending more per transaction.

* Seasonality analysis reveals recurring high-activity months, indicating predictable periods of increased demand that could be strategically leveraged.

* Average Order Value (AOV) remains relatively stable over time, confirming that the platform follows a high-volume, low-ticket sales model.



# **Recommendations & Next Steps (ACT)**

Based on the analysis of customer behavior, revenue distribution, and time-based trends, the following actions and next steps are recommended:

**1. Focus on Customer Retention Strategies**

The dominance of one-time buyers indicates low customer retention. Introducing retention-focused initiatives such as:

    * Loyalty or rewards programs
    
    * Personalized follow-up communications
    
    * Targeted discounts for returning customers

could help convert one-time buyers into repeat customers and improve long-term customer value.

**2. Leverage Seasonal Peaks for Growth**

Clear peak months and seasonal demand patterns were identified.The business can capitalize on these periods by:
    
    * Running targeted marketing campaigns ahead of peak months
    
    * Ensuring sufficient inventory during high-demand periods
    
    * Aligning promotions and pricing strategies with seasonal trends

**3. Increase Average Order Value (AOV)**

Since AOV remains relatively stable, revenue growth is largely volume-driven.
To improve per-order revenue:

    * Product bundling strategies
    
    * Cross-selling and upselling recommendations
    
    * Minimum order value incentives for free shipping

can be explored.

**4. Reduce Over-Reliance on New Customer Acquisition**

While revenue is not concentrated among a small set of customers (low dependency risk), the business currently relies heavily on acquiring new customers for growth.
Balancing acquisition efforts with retention-focused investments may lead to more sustainable growth.

**5. Opportunities for Further Analysis**

Future analyses could expand this work by exploring:

    * Product and category-level performance

    * Seller-level contribution and concentration

    * Payment behavior and installment usage

    * Delivery performance using additional order timestamps

These extensions could provide deeper operational and strategic insights.
